```python
import os
import pandas as pd
from numpy import array
from numpy import diag
from numpy import zeros
from scipy.linalg import svd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame(
    np.array([
        [7, 6.5, 9.2, 8.6, 8],
        [7.5, 9.4, 7.3, 7, 7],
        [7.6, 9.2, 8, 8, 7.5],
        [5, 6.5, 6.5, 7, 9],
        [6, 6, 7.8, 8.9, 7.3],
        [7.8, 9.6, 7.7, 8, 6.5],
        [6.3, 6.4, 8.2, 9, 7.2],
        [7.9, 9.7, 7.5, 8, 6],
        [6, 6, 6.5, 5.5, 8.7],
        [6.8, 7.2, 8.7, 9, 7]
    ]),
    index=['Lucia', 'Pedro', 'Ines', 'Luis', 'Andres',
        'Ana', 'Carlos', 'Jose', 'Sonia', 'Maria'],
    columns=[ 'Matematicas', 'Ciencias', 'Español', 'Historia', 'EdFisica'])

df 
```

```python
etiquetas = df.index
etiquetas
A = df
A
```

```python
#Aplicar el algoritmo SVD
U, s, VT = svd(A)

Sigma = zeros((A.shape[0], A.shape[1]))

Sigma[:A.shape[1], :A.shape[1]] = diag(s)
```

```python
print("Matriz U:")
print(U)
```

```python
print("Matriz Sigma")
print(Sigma)
```

```python
print("Matriz VT")
print(VT)
```

```python
#Reconstruir la matriz original
B = U.dot(Sigma.dot(VT))
print(B)
```

```python
#Reducción a dos dimensiones
n_elements = 2

UReduced2 = U[:, :n_elements]
print("U Reducida:")
print(UReduced2)
```

```python
#Reducir la matriz Sigma
SigmaReduced2 = Sigma[:n_elements, :n_elements]
print("Matriz Sigma Reducida")
print(SigmaReduced2)
```

```python
#Reducir la matriz VT
VTReduced2 = VT[:n_elements, :]
print("VT Reducida")
print(VTReduced2)
```

```python
#Obtener la matriz transformada
AReduced2 = UReduced2.dot(SigmaReduced2.dot(VTReduced2))
print("Matriz A transformada:")
print(AReduced2)
```

```python
#Obtener la matriz T de dos dimensiones
T2 = UReduced2.dot(SigmaReduced2)
print("Matriz T (2 Dimensiones):")
print(T2)
```

```python
#Graficar los resultados en dos dimensiones
df2 = pd.DataFrame(T2, columns=["a", "b"])

df2.plot(kind="scatter", x="a", y="b")
plt.show()
```

```python
df2 = pd.DataFrame(T2, columns=["a", "b"])

x = df2.iloc[:, 0]
y = df2.iloc[:, 1]

x = x.to_numpy()
y = y.to_numpy()

fig, ax = plt.subplots()
ax.set_title("Mapa de observaciones")
ax.scatter(x, y)

for i, txt in enumerate(etiquetas):
    ax.annotate(txt, (x[i], y[i]))

plt.show()
```

```python
#Reducción a tres dimensiones

n_elements = 3

UReduced3 = U[:, :n_elements]

SigmaReduced3 = Sigma[:n_elements, :n_elements]

VTReduced3 = VT[:n_elements, :]

AReduced3 = UReduced3.dot(SigmaReduced3.dot(VTReduced3))
print("Matriz A transformada:")
print(AReduced3)
```

```python
#Obtener la matriz T de tres dimensiones
T3 = UReduced3.dot(SigmaReduced3)
print("Matriz T (3 Dimensiones):")
print(T3)
```

```python
from mpl_toolkits import mplot3d
%matplotlib inline
```

```python
#Crear el DataFrame de tres dimensiones
df3 = pd.DataFrame(T3, columns=["a", "b", "c"])
df3
```

```python
#Graficar los resultados en tres dimensiones
fig = plt.figure()
ax = plt.axes(projection="3d")

xline = df3["b"]
yline = df3["c"]
zline = df3["a"]

ax.scatter3D(xline, yline, zline)

plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure()
ax3d = fig.add_subplot(111, projection="3d")

arrayx = np.array(df3["a"])
arrayy = np.array(df3["b"])
arrayz = np.array(df3["c"])

ax3d.scatter(arrayx, arrayy, arrayz)

for x, y, z, label in zip(arrayx, arrayy, arrayz, etiquetas):
    ax3d.text(x, y, z, label)

plt.title("Datos en 3D")
plt.show()
```

Calcular los errores de reconstrucción

```python
#Error de la reducción a dos dimensiones
plt.figure(figsize=(10, 7))

Error2 = B - AReduced2

dfError2 = pd.DataFrame(
    Error2,
    columns=["Matematicas", "Ciencias", "Español",
             "Historia", "EdFisica"]
)

dfError2.boxplot()
plt.show()
```

```python
print("Errores promedio por columna")
Error2.mean(axis=0)
```

```python
#Error de reduccion a tres dimensiones
Error3 = B - AReduced3
print(Error3)
```

```python
plt.figure(figsize=(10, 7))

dfError3 = pd.DataFrame(
    Error3,
    columns=["Matematicas", "Ciencias", "Español",
             "Historia", "EdFisica"]
)

dfError3.boxplot()
plt.show()
```

```python
print("Errores promedio por columna")
Error3.mean(axis=0)
```

```python
#Comparar las reducciones
print("Error de reconstrucción con 2 dimensiones:")
print(np.linalg.norm(Error2))

print("Error de reconstrucción con 3 dimensiones:")
print(np.linalg.norm(Error3))
```

*Interpretación de resultados*

Al aplicar el algoritmo SVD, logramos reducir las cinco variables originales a dos y tres dimensiones, lo que nos permitió representar gráficamente a los estudiantes y observar sus diferencias y similitudes en las calificaciones.

Con la reducción a dos dimensiones podemos identificar estudiantes que presentan comportamientos académicos similares, mientras que con tres dimensiones tenemos una representación más completa de las diferencias entre ellos.

Al comparar los errores de reconstrucción, encontramos que la reducción a tres dimensiones presenta un error menor (1.5449) que la reducción a dos dimensiones (4.1072). Esto indica que la representación tridimensional conserva mejor la información original.
